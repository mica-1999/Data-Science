# 🧠 4.1 k-Nearest Neighbors (kNN) From Scratch

## 📌 Objective

The first machine learning model implemented in this phase was a custom k-Nearest Neighbors (kNN) algorithm developed entirely from scratch using NumPy arrays. The purpose of this implementation was to understand the internal mechanics of distance-based learning instead of relying on the built-in `scikit-learn` implementation.

The kNN algorithm was applied to both:
- 📈 Regression: predicting the numerical arrival delay (`ARR_DELAY`)
- 🏷️ Classification: predicting delay categories (on-time, short delay, long delay)

---

## ⚙️ kNN Algorithm Overview

The k-Nearest Neighbors algorithm is a non-parametric and instance-based learning method. Instead of learning explicit model parameters during training, the algorithm stores the training data and makes predictions based on the similarity between observations.

For each test sample:
1. The Euclidean distance to all training samples is calculated
2. The `k` closest neighbors are selected
3. Predictions are generated:
   - Regression → average of neighbor target values
   - Classification → majority vote among neighbor classes

The Euclidean distance metric used in this implementation is:

```math
d(x,y)=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
```

---

## 📊 Evaluation Strategy

Multiple values of `k` were tested in order to evaluate how neighborhood size affects model performance. The tested values were configured through the YAML configuration file, allowing flexible experimentation without modifying the source code directly.

The following `k` values were evaluated:

```yaml
k_values: [13, 15, 17, 19, 21, 23, 25, 27, 29]
```

Higher `k` values generally produce smoother predictions and reduce sensitivity to noise, while lower values can capture local patterns more aggressively but may increase overfitting. The selected range was intended to balance stability and predictive performance across both regression and classification tasks.

The following evaluation metrics were used:

### Regression Metrics
- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)
- R² Score

### Classification Metrics
- Accuracy
- Precision
- Recall
- F1-score
- Confusion Matrix

---

````python
import yaml
import pandas as pd
import contextlib
import io

from ProjectCodes.PythonProject.src.kNN import KNNRunner

with open("config.yaml", "r") as f:
    config = yaml.safe_load(f)

knn_runner = KNNRunner(df, config=config)

f = io.StringIO()

with contextlib.redirect_stdout(f):
    regression_results = knn_runner.run_regression()
    classification_results = knn_runner.run_classification()

regression_df = pd.DataFrame(regression_results)
classification_df = pd.DataFrame(classification_results)

print("kNN Regression Results")
display(regression_df)

print("kNN Classification Results")
display(classification_df)
````

## 💬 kNN Results Discussion

The regression results showed a gradual improvement as the value of `k` increased. Larger neighborhood sizes produced lower MAE and RMSE values, indicating that averaging over more neighboring samples reduced sensitivity to noise and outliers in the dataset.

For the regression task, higher values of `k` were used (`13–29`) because larger neighborhood sizes generally produce more stable continuous predictions by reducing local variance.

The best regression performance was achieved with `k=29`, producing:

* MAE ≈ 20.16
* RMSE ≈ 45.82
* R² ≈ -0.031

Although prediction error improved slightly with larger values of `k`, all regression models still produced negative R² scores. This indicates that the custom kNN regressor performed worse than a simple baseline predictor using the mean arrival delay. These results suggest that the flight delay dataset contains complex nonlinear relationships and substantial variability that are difficult to capture using a basic distance-based approach.

For the classification task, a separate range of smaller `k` values (`3–15`) was tested, since classification problems typically benefit from smaller neighborhood sizes that preserve local class boundaries more effectively.

The best classification performance was achieved with `k=3`, producing:

* Accuracy ≈ 47.3%
* Macro F1 ≈ 47.4%

Classification performance gradually decreased as the value of `k` increased, suggesting that excessive smoothing between neighborhoods reduced the model’s ability to distinguish between delay categories.

These results are also consistent with guidance provided by the professor on Discord, where classification accuracies around 50% were discussed as realistic for this dataset and problem formulation. The obtained results therefore align with the expected difficulty of predicting flight delay categories using only pre-departure information.

Overall, the custom kNN implementation successfully demonstrated the mechanics of instance-based learning, Euclidean distance computation, regression, and multi-class classification using NumPy only. However, the model still struggled with the scale, dimensionality, and complexity of the flight delay dataset, motivating the use of more advanced supervised, ensemble, and deep learning methods in later sections.

## 📌 Objective

Following the custom kNN implementation, additional supervised learning algorithms were evaluated using the `scikit-learn` framework. These models were selected to compare different learning strategies and assess their ability to capture nonlinear relationships within the flight delay dataset.

The supervised learning phase included:
- 🌲 Decision Tree Regressor
- 📏 Support Vector Regressor (SVR)

Both models were applied to the regression task of predicting numerical arrival delay duration (`ARR_DELAY`).

---

## ⚙️ Model Overview

### 🌲 Decision Tree Regressor

Decision Trees partition the feature space into smaller regions by recursively splitting the data according to feature thresholds. The model learns decision rules that minimize prediction error while capturing nonlinear relationships and feature interactions.

Advantages:
- Interpretable structure
- Handles nonlinear patterns
- Does not require feature scaling

Limitations:
- Sensitive to overfitting
- High variance on noisy datasets

---

### 📏 Support Vector Regressor (SVR)

Support Vector Regression attempts to fit a function within a specified error margin while maximizing generalization performance. SVR is capable of modeling nonlinear relationships using kernel transformations.

Advantages:
- Effective for nonlinear regression
- Strong theoretical generalization properties
- Robust to high-dimensional spaces

Limitations:
- Computationally expensive on large datasets
- Sensitive to parameter tuning and feature scaling

---

## 📊 Evaluation Strategy

The supervised learning models were evaluated using standard regression metrics:

- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)

# 🌲 4.3 Ensemble Learning Models

## 📌 Objective

Following the supervised learning models, ensemble learning techniques were implemented to improve predictive performance and model robustness. Ensemble methods combine multiple weaker learners in order to produce more stable and accurate predictions than individual models alone.

The ensemble learning phase included:
- 🌳 Random Forest Regressor
- 🚀 Gradient Boosting Regressor

Both models were applied to the regression task of predicting numerical arrival delay duration (`ARR_DELAY`).

---

## ⚙️ Model Overview

### 🌳 Random Forest Regressor

Random Forest is a bagging-based ensemble method that combines predictions from multiple Decision Trees trained on random subsets of the data and features. The final prediction is obtained by averaging the outputs of all trees.

Advantages:
- Reduces overfitting compared to a single Decision Tree
- Handles nonlinear relationships effectively
- Robust to noise and outliers
- Provides feature importance analysis

Limitations:
- Less interpretable than a single tree
- Higher computational cost

---

### 🚀 Gradient Boosting Regressor

Gradient Boosting builds trees sequentially, where each new tree attempts to correct the prediction errors of the previous ensemble. This boosting strategy allows the model to gradually improve predictive accuracy.

Advantages:
- Strong predictive performance
- Captures complex nonlinear patterns
- Effective on structured/tabular datasets

Limitations:
- Sensitive to hyperparameter tuning
- More prone to overfitting if boosting is too aggressive
- Slower training compared to simpler models

---

## 📊 Evaluation Strategy

The ensemble learning models were evaluated using the following regression metrics:

- Mean Absolute Error (MAE)
- Root Mean Squared Error (RMSE)
- R² Score

Model hyperparameters such as the number of estimators, tree depth, and learning rate were configured through the YAML configuration file, allowing flexible experimentation without directly modifying the implementation code.

# 🔍 4.5 Clustering Analysis

## 📌 Objective

The final model-building task was clustering, which was used to identify hidden operational patterns in the flight delay dataset. Unlike the previous models, clustering is an unsupervised learning approach, meaning that it does not use a target variable for prediction.

The goal of this section was to group flights based on operational characteristics such as delay behavior, distance, scheduled flight duration, departure time, and engineered features. These clusters can help reveal patterns related to airline punctuality, airport congestion, route behavior, and general operational performance.

---

## ⚙️ Clustering Models

Two clustering algorithms were applied:

### 🔵 KMeans Clustering

KMeans is a centroid-based clustering algorithm that partitions observations into a predefined number of clusters. The algorithm attempts to minimize the distance between each observation and the center of its assigned cluster.

In this project, multiple values of `k` were tested to evaluate whether meaningful cluster structures existed in the data.

Advantages:
- Simple and efficient
- Useful for identifying broad operational groupings
- Works well when clusters are relatively compact and separated

Limitations:
- Requires choosing the number of clusters
- Sensitive to feature scaling
- Assumes roughly spherical cluster shapes

---

### 🟣 DBSCAN Clustering

DBSCAN is a density-based clustering algorithm that groups observations based on local point density. Unlike KMeans, it does not require specifying the number of clusters beforehand and can identify noise points or outliers.

Advantages:
- Can detect irregular cluster shapes
- Identifies noise and anomaly-like observations
- Does not require a fixed number of clusters

Limitations:
- Sensitive to `eps` and `min_samples`
- Can struggle when data density varies strongly
- May classify many points as noise depending on parameter settings

---

## 📊 Evaluation Strategy

The clustering models were evaluated using:

- Inertia for KMeans
- Silhouette Score for cluster separation
- Cluster counts
- DBSCAN number of clusters
- DBSCAN noise ratio
- PCA-based visualizations of cluster assignments

The number of clusters and DBSCAN parameters were configured through the YAML configuration file, allowing experimentation without directly modifying the implementation code.

---

````python
import yaml
import pandas as pd
import contextlib
import io

from ProjectCodes.PythonProject.src.clustering import ClusteringRunner

with open("config.yaml", "r") as f:
    config = yaml.safe_load(f)

clustering_runner = ClusteringRunner(df, config=config)

f = io.StringIO()

with contextlib.redirect_stdout(f):
    clustering_runner.run_all()

kmeans_results = pd.read_csv(config["output_dir_model_results"] + "/clustering_kmeans_results.csv")
dbscan_results = pd.read_csv(config["output_dir_model_results"] + "/clustering_dbscan_results.csv")
kmeans_counts = pd.read_csv(config["output_dir_model_results"] + "/clustering_kmeans_cluster_counts.csv")
dbscan_counts = pd.read_csv(config["output_dir_model_results"] + "/clustering_dbscan_cluster_counts.csv")

print("KMeans Results")
display(kmeans_results)

print("DBSCAN Results")
display(dbscan_results)

print("KMeans Cluster Counts")
display(kmeans_counts)

print("DBSCAN Cluster Counts")
display(dbscan_counts)
````

## 💬 Clustering Results Discussion

The clustering analysis identified several operational groupings within the flight delay dataset using both centroid-based and density-based clustering methods.

---

### 🔵 KMeans Results

KMeans clustering was evaluated using values of `k` ranging from 2 to 8.

The highest silhouette score was achieved at:

- `k = 2`
- Silhouette Score ≈ 0.246

As the number of clusters increased, inertia continued to decrease, which is expected because additional clusters reduce within-cluster variance. However, the silhouette scores generally decreased or remained relatively low, suggesting that the dataset does not contain strongly separated cluster structures.

The final KMeans clustering produced two dominant operational groups:
- Cluster 0 → approximately 36,529 flights
- Cluster 1 → approximately 63,471 flights

These clusters likely represent broad operational behaviors such as:
- lower-delay vs higher-delay operational patterns
- short-haul vs long-haul flight structures
- different scheduling and traffic profiles

The relatively modest silhouette scores indicate that operational flight behavior overlaps significantly across flights, airlines, and routes.

---

### 🟣 DBSCAN Results

DBSCAN clustering produced:
- 7 detected clusters
- approximately 2.98% noise points

Most observations were concentrated in one large dense cluster:
- Cluster 0 → approximately 92,297 flights

Several much smaller clusters and anomaly-like groups were also detected. These likely correspond to unusual operational behaviors, rare delay patterns, or less common route and scheduling combinations.

The presence of noise points demonstrates one advantage of DBSCAN compared to KMeans: the ability to identify observations that do not belong clearly to any dense operational pattern.

---

### 📌 Overall Interpretation

The clustering analysis revealed that the flight delay dataset contains broad operational structures but relatively weak cluster separation overall. This is reasonable because flight operations are influenced by many overlapping factors such as:
- airline schedules
- airport congestion
- weather conditions
- route distance
- seasonal effects

Although the clusters were not strongly separated, the analysis still provided insight into underlying operational behaviors and demonstrated how unsupervised learning can be used to explore patterns in airline and airport performance without predefined labels.
